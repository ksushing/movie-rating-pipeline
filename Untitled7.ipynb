```python
import pandas as pd
import os
import numpy as np

np.random.seed(42)


df = pd.read_csv('/content/movies.csv')

df = df.sample(frac=1, random_state=42).reset_index(drop=True)


n = len(df)
n_initial = int(n * 0.6)
n_batches = 5
n_batch = (n - n_initial) // n_batches

os.makedirs('data/raw', exist_ok=True)
os.makedirs('data/incoming', exist_ok=True)

initial = df.iloc[:n_initial]
initial.to_csv('data/raw/movies_initial.csv', index=False)


for i in range(n_batches):
    start = n_initial + i * n_batch
    end = start + n_batch if i < n_batches - 1 else n
    batch = df.iloc[start:end]
    batch.to_csv(f'data/incoming/batch_{i+1:02d}.csv', index=False)
    print(f'batch_{i+1:02d}.csv: {len(batch)} строк')

print(f'Исходная часть: {len(initial)} строк')
print(f'Всего: {n} строк')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

%matplotlib inline
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)


df = pd.read_csv('data/raw/movies_initial.csv')

print("Размер исходной части:", df.shape)

#Структура набора

print("\nТипы данных:")
print(df.dtypes)

print("\nПропуски:")
missing = df.isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Пропуски': missing, 'Процент': missing_pct})
print(missing_df[missing_df['Пропуски'] > 0])

print("\nДубликаты:", df.duplicated().sum())

#Целевая переменная

print("\nСтатистика score:")
print(df['score'].describe())

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist(df['score'].dropna(), bins=30, color='steelblue', edgecolor='black')
axes[0].set_title('Распределение score')
axes[0].set_xlabel('score')

axes[1].boxplot(df['score'].dropna())
axes[1].set_title('Boxplot score')
plt.tight_layout()
plt.show()

#Качество данных

# Выбросы в числовых признаках
numeric_cols = ['score', 'votes', 'budget', 'gross', 'runtime', 'year']
numeric_cols = [c for c in numeric_cols if c in df.columns]

print("\nВыбросы по IQR:")
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)]
    print(f"{col}: {len(outliers)} выбросов ({len(outliers)/len(df)*100:.1f}%)")

#Связь с целевой переменной

print("\nКорреляция с score:")
corr = df[numeric_cols].corr()['score'].sort_values(ascending=False)
print(corr)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for i, col in enumerate(['votes', 'runtime', 'budget']):
    axes[i].scatter(df[col], df['score'], alpha=0.3, color='steelblue')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('score')
    axes[i].set_title(f'{col} vs score')
plt.tight_layout()
plt.show()

#Взаимосвязи признаков

plt.figure(figsize=(10, 8))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Корреляционная матрица')
plt.show()

#Признаки с утечкой

print("\nПризнаки, потенциально содержащие утечку:")
print("- votes: известен только после выхода фильма")
print("- gross: известен только после проката")

#ГИПОТЕЗЫ

print("\n" + "=" * 60)
print("ПРОВЕРКА ГИПОТЕЗ")
print("=" * 60)

# Гипотеза 1: votes → score
r, p = stats.pearsonr(df['votes'].dropna(), df.loc[df['votes'].notna(), 'score'])
print(f"\nГипотеза 1: votes и score")
print(f"Корреляция Пирсона: r = {r:.3f}, p = {p:.4f}")
print("Вывод:", "подтверждена" if p < 0.05 and abs(r) > 0.3 else "не подтверждена")

# Гипотеза 2: genre → score (ANOVA)
groups = [g['score'].dropna() for name, g in df.groupby('genre')]
f_stat, p_value = stats.f_oneway(*groups)
print(f"\nГипотеза 2: genre и score (ANOVA)")
print(f"F = {f_stat:.3f}, p = {p_value:.4f}")
print("Вывод:", "подтверждена" if p_value < 0.05 else "не подтверждена")

# Гипотеза 3: budget → score
subset = df[['budget', 'score']].dropna()
r, p = stats.pearsonr(subset['budget'], subset['score'])
print(f"Гипотеза 3: budget и score")
print(f"Корреляция Пирсона: r = {r:.3f}, p = {p:.4f}")
print(f"Размер выборки: {len(subset)}")
print("Вывод:", "подтверждена" if p < 0.05 and abs(r) > 0.3 else "не подтверждена")
```

```python
import pandas as pd

def check_quality(df, name="датасет"):
    """Проверка требований к качеству данных."""
    print(f"\nПроверка: {name}")
    print(f"Строк: {len(df)}")

    # 1. name + year уникальны
    dups = df.duplicated(subset=['name', 'year']).sum()
    print(f"[1] Дубликаты name+year: {dups} " + ("OK" if dups == 0 else "НАРУШЕНИЕ"))

    # 2. score в диапазоне 0..10 и не пуст
    bad_score = df[(df['score'].isnull()) | (df['score'] < 0) | (df['score'] > 10)]
    print(f"[2] score вне 0..10 или пуст: {len(bad_score)} " + ("OK" if len(bad_score) == 0 else "НАРУШЕНИЕ"))

    # 3. votes >= 0 и не пуст
    bad_votes = df[(df['votes'].isnull()) | (df['votes'] < 0)]
    print(f"[3] votes < 0 или пуст: {len(bad_votes)} " + ("OK" if len(bad_votes) == 0 else "НАРУШЕНИЕ"))

    # 4. budget >= 0 или пропуск
    bad_budget = df[(df['budget'].notnull()) & (df['budget'] < 0)]
    print(f"[4] budget < 0: {len(bad_budget)} " + ("OK" if len(bad_budget) == 0 else "НАРУШЕНИЕ"))

    # 5. runtime в диапазоне 1..500
    bad_runtime = df[(df['runtime'].notnull()) & ((df['runtime'] < 1) | (df['runtime'] > 500))]
    print(f"[5] runtime вне 1..500: {len(bad_runtime)} " + ("OK" if len(bad_runtime) == 0 else "НАРУШЕНИЕ"))

    # 6. year в диапазоне 1900..2026
    bad_year = df[(df['year'] < 1900) | (df['year'] > 2026)]
    print(f"[6] year вне 1900..2026: {len(bad_year)} " + ("OK" if len(bad_year) == 0 else "НАРУШЕНИЕ"))


# Проверка исходной части
initial = pd.read_csv('data/raw/movies_initial.csv')
check_quality(initial, "movies_initial.csv")

# Проверка порций
for i in range(1, 6):
    batch = pd.read_csv(f'data/incoming/batch_{i:02d}.csv')
    check_quality(batch, f"batch_{i:02d}.csv")
```

```python
import pandas as pd
import numpy as np

# Загружаем очищенную исходную часть
df = pd.read_csv('data/raw/movies_initial_clean.csv')

# 1. Бюджет на минуту (чтобы не было деления на ноль, добавим 1)
df['budget_per_minute'] = df['budget'] / (df['runtime'] + 1)

# 2. Голоса в год (предполагаем, что текущий год - 2024)
df['votes_per_year'] = df['votes'] / (2024 - df['year'] + 1)

# 3. Флаг англоязычного фильма
df['is_english'] = df['country'].apply(lambda x: 1 if x in ['United States', 'United Kingdom'] else 0)

# 4. Десятилетие выхода фильма
df['decade'] = (df['year'] // 10) * 10

# Выводим результат, чтобы сделать скриншот для отчета
new_cols = ['name', 'year', 'budget', 'runtime', 'budget_per_minute',
            'votes', 'votes_per_year', 'country', 'is_english', 'decade']

print("Пример сгенерированных признаков:")
print(df[new_cols].head())

# Проверяем, что не появилось бесконечностей или ошибок
print("\nПроверка на NaN в новых признаках:")
print(df[['budget_per_minute', 'votes_per_year', 'is_english', 'decade']].isnull().sum())

# Сохраняем датасет с новыми признаками (в реальном проекте это ушло бы в data/processed)
df.to_csv('data/raw/movies_with_features.csv', index=False)
print("\nСохранено: data/raw/movies_with_features.csv")
```
